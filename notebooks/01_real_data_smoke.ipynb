# 01 Real Data Smoke

This notebook uses **DataHub** to inspect source capability coverage table output, discover assets, ingest market datasets, and review source issues.


```python
from algotradeplan.data import DataHub

hub = DataHub()

sources = hub.sources()
coverage = hub.coverage_table()
assets = hub.discover_assets(source="binance_futures", limit=5)
result = hub.ingest(
    source="binance_futures",
    symbol=assets[0] if assets else "BTCUSDT",
    datasets=["kline", "trade", "orderbook", "funding"],
    timeframe="1m",
    limit=120,
    allow_partial=True,
)

{
    "sources": sources[:5],
    "coverage_preview": coverage[:2],
    "assets": assets,
    "dataset_coverage": result.dataset_coverage,
    "data_quality": result.quality_report,
    "source issues": result.source_issues,
}
```
