# 01 Real Data Smoke

This notebook uses **DataHub** for coverage table preview, multi-source ingest examples, and source issues inspection.


```python
from algotradeplan.data import DataHub

hub = DataHub()
coverage = hub.coverage_table()
sources = hub.sources()
assets = hub.discover_assets(source="binance_futures", limit=5)

market = hub.ingest(source="binance_futures", symbol=(assets[0] if assets else "BTCUSDT"), datasets=["kline", "trade"], allow_partial=True)
news = hub.ingest(source="gdelt", symbol="bitcoin", datasets=["news"], allow_partial=True)
macro = hub.ingest(source="world_bank", symbol="NY.GDP.MKTP.CD", datasets=["macro"], allow_partial=True)

{
    "sources": sources[:8],
    "assets": assets,
    "coverage_table": coverage[:5],
    "market_dataset_coverage": market.dataset_coverage,
    "news_dataset_coverage": news.dataset_coverage,
    "macro_dataset_coverage": macro.dataset_coverage,
    "source issues": market.source_issues + news.source_issues + macro.source_issues,
}
```
